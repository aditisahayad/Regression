Question 1: What is Simple Linear Regression?

Answer 1:Simple Linear Regression is a statistical method that models the relationship between two variables: one independent variable (X) and one dependent variable (Y). It fits a straight line to the data using the equation: **Y=mX+c**

where:
- Y is the predicted value,
- X is the input variable,
- m is the slope,
- c is the intercept.

Question 2: What are the key assumptions of Simple Linear Regression?

Answer 2: Key assumptions of Simple Linear Regression:

- Linearity: The relationship between X and Y is linear.

- Independence: Observations are independent of each other.

- Homoscedasticity: Constant variance of residuals (errors).

- Normality: Residuals should be normally distributed.

- No multicollinearity (though this mainly applies to multiple regression).

Question 3: What does the coefficient
m represent in the equation
Y = mX+c?

Answer 3:
The coefficient m represents the slope of the regression line. It indicates the change in the dependent variable Y for a one-unit increase in the independent variable X.

Question 4: What does the intercept c represent in the equation Y=mX+c?

Answer 4:
The intercept c is the value of Y when X=0. It's where the regression line crosses the Y-axis.

Question 6: What is the purpose of the least squares method in Simple Linear Regression?

Answer 6:
The least squares method minimizes the sum of the squared differences between the actual values and the predicted values. This helps find the best-fitting line through the data.

Question 7: How is the coefficient of determination (R^2) interpreted in Simple Linear Regression?

Answer 7:
(R^2)measures the proportion of the variance in the dependent variable that is predictable from the independent variable. An (R^2) of 0.8 means 80% of the variance in Y is explained by X.

Question 8: What is Multiple Linear Regression?

Answer 8:
Multiple Linear Regression models the relationship between one dependent variable and two or more independent variables.2.jpg

Question 9: What is the main difference between Simple and Multiple Linear Regression?

Answer 9:
Simple Linear Regression uses one independent variable, while Multiple Linear Regression uses two or more independent variables to predict the dependent variable.

Question 10: What are the key assumptions of Multiple Linear Regression?

Answer 10: Key assumptions of Multiple Linear Regression:
- Linearity between independent and dependent variables.
- Independence of errors.
- Homoscedasticity (constant variance of errors).
- Normal distribution of errors.
- No or little multicollinearity among independent variables.

Question 11: What is heteroscedasticity, and how does it affect the results of a Multiple Linear Regression model?

Answer 11:
Heteroscedasticity occurs when the variance of the residuals is not constant across all levels of the independent variables. It can make standard errors unreliable and affect confidence intervals and hypothesis tests.

Question 12: How can you improve a Multiple Linear Regression model with high multicollinearity?

Answer 12:

- Remove or combine correlated variables.
- Use dimensionality reduction (e.g., PCA).
- Apply regularization techniques like Ridge or Lasso Regression.



Question 13: What are some common techniques for transforming categorical variables for use in regression models?

Answer 13:
- One-hot encoding: Creates binary columns for each category.

- Label encoding: Converts categories into integers.

- Ordinal encoding: Assigns ordered integers to ordinal categories.



Question 14: What is the role of interaction terms in Multiple Linear Regression?

Answer 14:
Interaction terms capture the effect of two variables acting together, showing how the impact of one variable on the dependent variable changes depending on the level of another variable.

Question 15: How can the interpretation of intercept differ between Simple and Multiple Linear Regression?

Answer 15:
In Simple Linear Regression, the intercept is the value of Y when X = 0.
In Multiple Linear Regression, the intercept is the predicted Y when all X variables are 0, which may not always be meaningful in real-world context

Question 16: What is the significance of the slope in regression analysis, and how does it affect predictions?

Answer 16:
The slope shows how much the dependent variable is expected to increase (or decrease) when an independent variable increases by one unit. It directly influences predictions.



Question 17: How does the intercept in a regression model provide context for the relationship between variables?

Answer 17:
The intercept provides a baseline value for the dependent variable when all independent variables are zero. It sets the starting point for the prediction.

Question 18: What are the limitations of using (R^2) as a sole measure of model performance?

Answer 18:

- It doesn't indicate whether predictions are biased.

- It doesn't reveal overfitting.

- Adding more variables can artificially increase (R^2) even if they aren't meaningful.

Question 19: How would you interpret a large standard error for a regression coefficient?

Answer 19:
A large standard error suggests high uncertainty in the estimate of that coefficient, meaning it may not be a reliable predictor.

Question 20: How can heteroscedasticity be identified in residual plots, and why is it important to address it?

Answer 20:
Heteroscedasticity is identified when residual plots show a fan or cone shape instead of a random scatter. Addressing it is important because it can lead to incorrect inferences in regression results.

Question 21: What does it mean if a Multiple Linear Regression model has a high (R^2) but low adjusted (R^2)?

Answer 21:
It suggests that the model includes variables that don't improve predictive power. Adjusted (R^2) penalizes unnecessary variables, unlike (R^2).



Question 22: Why is it important to scale variables in Multiple Linear Regression?

Answer 22:
- Scaling ensures all variables are on the same scale, which helps:

- Improve model convergence.

- Prevent bias in coefficients.

Enhance interpretation, especially when using regularization.



Question 23: What is polynomial regression?

Answer 23:
Polynomial Regression is a form of regression where the relationship between independent and dependent variable is modeled as an n-th degree polynomial.

Question 24: How does polynomial regression differ from linear regression?

Answer 24:
Linear regression models a straight-line relationship, while polynomial regression models curved (non-linear) relationships using higher-degree terms of the input variable.

Question 25: When is polynomial regression used?

Answer 25:
It's used when the data shows a curvilinear trend that cannot be captured by a straight line, and you suspect a nonlinear relationship between variables.
